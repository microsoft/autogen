# Monetizing AI Agents with the x402 Protocol (Base Network)

In this tutorial, we will learn how to give our autonomous AI agent a crypto wallet and let it pay for its own tools and data using the **x402 protocol** on the Base network.

We will connect to the `x402-agent-tollbooth` MCP server, which requires agents to pay USDC per-call to execute smart contract audits and token analytics.

### Setting up the Agent Wallet
Your agent needs a funded wallet on Base to execute micro-transactions.

```
from web3 import Web3
import os

# Connect to Base
w3 = Web3(Web3.HTTPProvider('https://mainnet.base.org'))

# Load your agent's private key
AGENT_PRIVATE_KEY = os.getenv('AGENT_PRIVATE_KEY', '0x...')
if AGENT_PRIVATE_KEY != '0x...':
    account = w3.eth.account.from_key(AGENT_PRIVATE_KEY)
    print(f'Agent Wallet Address: {account.address}')
```

### Connecting the MCP Tollbooth
We configure LangChain (or your framework of choice) to load tools from the remote MCP server.

```
# Pseudo-code for connecting to the Tollbooth MCP Server
# The agent will automatically call this when it needs to audit a contract.
TOLLBOOTH_MCP_URL = "https://glama.ai/mcp/servers/ak10082247-max/x402-agent-tollbooth"

print("MCP Tollbooth connected. Available tools: ['contract.audit', 'wallet.profile', 'token.analyze']")
```

### Execution
When the agent decides to use the `contract.audit` tool, it must first execute a transaction on Base to pay the 1.00 USDC toll, and pass the `paymentHash` into the tool call.
